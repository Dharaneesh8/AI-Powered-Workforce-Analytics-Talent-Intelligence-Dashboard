```python
import pandas as pd
import numpy as np
```

```python
# Generate sample e-commerce transaction data
np.random.seed(42) # for reproducibility

data = {
    'transaction_id': [f'TR{i:04d}' for i in range(1, 11)],
    'customer_id': np.random.randint(1000, 5000, 10),
    'product_id': [f'PRD{i:03d}' for i in np.random.randint(100, 999, 10)],
    'product_name': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam', 'Speaker', 'Headphones', 'Tablet', 'Smartphone', 'Smartwatch'],
    'category': np.random.choice(['Electronics', 'Accessories', 'Peripherals'], 10),
    'price': np.round(np.random.uniform(20, 1500, 10), 2),
    'quantity': np.random.randint(1, 5, 10),
    'order_date': pd.to_datetime(pd.date_range(start='2023-01-01', periods=10, freq='D')),
    'payment_method': np.random.choice(['Credit Card', 'PayPal', 'Bank Transfer'], 10),
    'shipping_cost': np.round(np.random.uniform(5, 50, 10), 2)
}

df_transactions = pd.DataFrame(data)

# Display the first 10 rows of the DataFrame
display(df_transactions)
```
